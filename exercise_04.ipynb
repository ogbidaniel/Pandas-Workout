## Descriptive Statistics

- Generate a series of 100,000 floats in a normal distribution with a mean of 0 and a standard deviation of 100.
- Get the descriptive statistics for this series.
- Replace the minimum value with 5 times the maximu value.

```python
import pandas as pd
import numpy as np
```

Series of 100_000 floats in a normal distribution

```python
g = np.random.default_rng(0)
s = pd.Series(g.normal(0, 100, 100_000))
```

Getting the descriptive stats with .describe()

```python
s.describe()
```

replace min val with max val * 5

```python
s.loc[s == s.min()]
```

```python
s.loc[s == s.min()] = 5*s.max()
s.describe()
```

## Beyond Excercises

Demonstrate that 68%, 95%, and 99.7% of the values in s are indeed within one, two, and three standard distributions of the mean.


```python
s.std()
```

```python
# 68% within 1 std?
within_1_std = ((s.loc[(s > (s.mean() - s.std())) & (s < (s.mean() + s.std()))].count()) / s.count()) * 100
within_1_std
```

```python
# 95% within 2 std
within_2_std = ((s.loc[(s > (s.mean() - 2*s.std())) & (s < (s.mean() + 2*s.std()))].count()) / s.count()) * 100
within_2_std
```

```python
# 99.7% within 3 std
within_3_std = ((s.loc[(s > (s.mean() - 3*s.std())) & (s < (s.mean() + 3*s.std()))].count()) / s.count()) * 100
within_3_std
```

Calculate the mean of numbers greater than s.mean(). Then calculate the mean of numbers less than s.mean(). Is the average of these two numbers the same as s.mean()?

```python
# mean of numbers greater than mean
highmean = s.loc[s > s.mean()].mean()
highmean
```

```python
# mean of numbers greater than mean
lowmean = s.loc[s < s.mean()].mean()
lowmean
```

```python
# is high mean and low mean mean same as regular mean?
testmean = (highmean + lowmean)/2
testmean == s.mean()
```
