```python
import pandas as pd
import numpy as np
from pandas import Series, DataFrame
```

```python
# load datasets
df_July_19 = pd.read_csv('data/nyc_taxi_2019-07.csv',
                        usecols=['passenger_count', 'total_amount', 'payment_type'])
df_July_19['year'] = 2019
df_July_20 = pd.read_csv('data/nyc_taxi_2020-07.csv',
                        usecols=['passenger_count', 'total_amount', 'payment_type'])
df_July_20['year'] = 2020
```

```python
# merge with concat
df = pd.concat([df_July_19, df_July_20])
df.head()
```

```python
# How many rides were taken in 2019 and 2020, and what is the difference between these two figures?
print(f"Total rides")
total2019 = df.loc[df['year']==2019, 'year'].count()
total2020 = df.loc[df['year']==2020, 'year'].count()
print(f"2019: {total2019}")
print(f"2020: {total2020}")
print(f"Difference: {total2020-total2019}")
```

```python
# How much money (in total) was collected in 2019 and 2020, and what was the difference between these two figures?
print(f"Total money")
total2019 = df.loc[df['year']==2019, 'total_amount'].sum()
total2020 = df.loc[df['year']==2020, 'total_amount'].sum()
print(f"2019: {total2019}")
print(f"2020: {total2020}")
print(f"Sum: {total2020+total2019}")
print(f"Difference: {total2020-total2019}")
```

```python
# Did the proportion of trips with more than one passenger change dramatically?
ratio19 = df.loc[(df['year'] == 2019) & (df['passenger_count'] > 1), 
'passenger_count'].count() / df.loc[df['year'] == 2019, 'payment_type'].count()
ratio20 = df.loc[(df['year'] == 2020) & (df['passenger_count'] > 1), 
'passenger_count'].count() / df.loc[df['year'] == 2020, 'payment_type'].count()

print(f"2019 ratio: {ratio19}")
print(f"2020 ratio: {ratio20}")
```

```python
# Did people use cash (i.e., payment_type of 2) less in 2020 than in 2019?
cash2019 = df.loc[(df['year'] == 2019) & (df['payment_type'] == 2),
'passenger_count'].count()
cash2020 = df.loc[(df['year'] == 2020) & (df['payment_type'] == 2),
'passenger_count'].count()
print(f"2019: {cash2019}")
print(f"2020: {cash2020}")
```
