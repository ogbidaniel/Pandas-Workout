# Exercise 8 - Net revenue

Create a data frame for a companies 5 inventory products. Each product has a unique ID number (a two-digit integer will do), name, wholesale price, retail price, and number of sales in the last month.

```python
import pandas as pd
import numpy as np
```

```python
# method 1
df = pd.DataFrame([
    {'product_id': 23, 'name': 'macbook air', 'wholesale_price': 500, 'retail_price': 1000, 'sales': 100},
    {'product_id': 96, 'name': 'Python Workout', 'wholesale_price': 35, 'retail_price':75, 'sales': 1000},
    {'product_id': 97, 'name': 'Pandas Workout', 'wholesale_price': 35, 'retail_price':75, 'sales': 500},
    {'product_id': 15, 'name': 'banana', 'wholesale_price': 0.5, 'retail_price':1, 'sales': 200},
    {'product_id': 87, 'name': 'sandwhich', 'wholesale_price': 3.0, 'retail_price':5, 'sales': 300},
])
```

```python
((df['retail_price'] - df['wholesale_price']) * df['sales']).sum()
```

## Beyond the exercise
