```python
import pandas as pd
from pandas import Series, DataFrame
```

```python
# Create the data frame based on the individual series.
"""
use `read_csv` to injest the csv file.
path to taxi distance.csv 'data/taxi-distance.csv'
path to taxi-passenger-count.csv 'data/taxi-distance.csv'
"""
# use absolute path.
trip_distance = pd.read_csv("C:/dev/Pandas-Workout/data/taxi-distance.csv", header=None).squeeze()
passenger_count = pd.read_csv("C:/dev/Pandas-Workout/data/taxi-passenger-count.csv", header=None).squeeze()

# use DataFrame
df = DataFrame({
    'trip_distance': trip_distance,
    'passenger_count':passenger_count
})
df.head(10)
```

```python
# iqr = df['trip_distance].quantile(0.75) - df['trip_distance'].quantile(0.25)
iqr = (
    df['trip_distance'].quantile(0.75) - 
    df['trip_distance'].quantile(0.25)
)
print(iqr)
```

```python
# outlier distances --below
df[
    df['trip_distance'] < 
    df['trip_distance'].quantile(0.25) - 
    1.5*iqr
]
```

```python
# outlier distances --above
df[df['trip_distance'] >
   df['trip_distance'].quantile(0.75) + 
   1.5 * iqr
]
```

```python
# mean passenger count for outliers
df['passenger_count'][
    df['trip_distance'] > df['trip_distance'].quantile(0.75) +
    1.5*iqr
].mean()
```

```python
df['passenger_count'].mean()
```

```python
# Beyond 2
"""
How many short, medium, and long trips were there for trips that had only one passenger? 
Note that data for passenger count and trip length are from the same data set, 
meaning that the indexes are the same.
"""
from scipy.stats import trimboth
trimboth(df['trip_distance'], 0.1)
```

```python
# Beyond 3
"""
The scipy.stats.zscore function rescales and centers (that is, normalizes) the data set. 
In this case, the mean is set to 0, and values can be above and below that value. 
Find all the distances for which the absolute value of the z-score is greater than 3.
"""
from scipy.stats import zscore
df['trip_distance'][abs(zscore(df['trip_distance'])) > 3]
```
