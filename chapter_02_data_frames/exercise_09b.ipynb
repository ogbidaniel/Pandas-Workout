Add sales tax of 15%, 20% and 25%

## Exercise

Add your summary of the task and any dataset notes here.

```python
import pandas as pd
import numpy as np
```

## Solution

```python
# method 1
df = pd.DataFrame([
    {'product_id': 23, 'name': 'macbook air', 'wholesale_price': 800, 'retail_price': 1000, 'sales': 40},
    {'product_id': 96, 'name': 'Python Workout', 'wholesale_price': 35, 'retail_price':75, 'sales': 1000},
    {'product_id': 97, 'name': 'Pandas Workout', 'wholesale_price': 35, 'retail_price':75, 'sales': 500},
    {'product_id': 15, 'name': 'banana', 'wholesale_price': 0.5, 'retail_price':1, 'sales': 300},
])
```

```python
df = df.assign(current_net = (df['retail_price'] - df['wholesale_price']) * df['sales'])
```

```python
# after 15% tax
df = df.assign(after_15 = (df['current_net'] * 0.85))
```

```python
# after 20% tax
df = df.assign(after_20 = df['current_net'] * 0.80)
```

```python
# after 25% tax
df = df.assign(after_25 = df['current_net'] * 0.75)
```

### Beyond 1

```python
df['current_net'].apply(lambda c: c*0.75 if c > 20000 else c).sum()
```

```python
# longer way
def calc_tax(c):
    if c > 20000:
        return c * 0.75
    else:
        return c

df['current_net'].apply(calc_tax).sum()
```

### beyond 2

```python
# 25% tax retail > 80 10% on 80 > retail > 30 and no tax on others
df['after_tax'] = pd.cut(df['retail_price'],
                         bins = [0, 30, 80, df['retail_price'].max()],
                         labels=[1, 0.9, 0.75]).astype(np.float64)

df['final_net'] = df['current_net'] - df['after_tax']
df
```
