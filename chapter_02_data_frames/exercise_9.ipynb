# Exercise 9: Tax Planning


Extend the existing inventory DataFrame by adding new columns that calculate the store's net income under different proposed sales tax rates.

1. Calculate the current net income for each product.
2. Create new columns representing net income after:
   - 15% tax
   - 20% tax
   - 25% tax
3. Compare the totals under each tax scenario to understand the financial impact.


```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame([{'product_id':23, 'name':'computer', 'wholesale_price': 500,
        'retail_price':1000, 'sales':100},
        {'product_id':96, 'name':'Python Workout', 'wholesale_price': 35,
        'retail_price':75, 'sales':1000},
        {'product_id':97, 'name':'Pandas Workout', 'wholesale_price': 35,
        'retail_price':75, 'sales':500},
        {'product_id':15, 'name':'banana', 'wholesale_price': 0.5,
        'retail_price':1, 'sales':200},
        {'product_id':87, 'name':'sandwich', 'wholesale_price': 3,
        'retail_price':5, 'sales':300},
        ])
```

```python
# Calculate current net income for each product
df['net_income'] = (df['retail_price'] - df['wholesale_price']) * df['sales']
```

```python
df['after_15'] = df['net_income'] * 0.85
df['after_20'] = df['net_income'] * 0.80
df['after_25'] = df['net_income'] * 0.75
```

```python
df
```

```python
print(f'Net income: {df['net_income'].sum()}')
print(f'Net income (after 15% tax): {df['after_15'].sum()}')
print(f'Net income (after 20% tax): {df['after_20'].sum()}')
print(f'Net income (after 25% tax): {df['after_25'].sum()}')
```
