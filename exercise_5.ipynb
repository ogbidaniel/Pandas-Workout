## Exercise 5 - Monday Temperatures

Create a series of 28 temperature readings in Celcius, representing 4 random 7-day weeks.
Random distrubution, mean of 20, standard deviation of 5. Rounded to nearest int.

```python
import pandas as pd
import numpy as np
from pandas import Series, DataFrame
```

```python
days = 'Sun Mon Tue Wed Thu Fri Sat'.split()
g = np.random.default_rng(0)
s = Series(g.normal(20, 5, 28),
           index=days*4).astype(np.int8)
s
```

what was the mean temperature on Mondays during this period?

```python
s['Mon'].mean()
```

## Beyond the Exercise

```python
# What was the average weekend temprature?
s[['Sat', 'Sun']].mean()
```

```python
# How many times is the change in temp from the previous day greater than 2 degrees?
s[s.diff() > 2].count()
```
