Show what percentage of taxi rides had only one passenger versus the (theoretical) maximum of six passengers

```python
import numpy as np
import pandas as pd
```

```python
pathtodata = '/Users/ogbi/Development/Pandas-Workout/data/taxi-passenger-count.csv'
s = pd.read_csv(pathtodata,
                header=None).squeeze()
s.head()
```
