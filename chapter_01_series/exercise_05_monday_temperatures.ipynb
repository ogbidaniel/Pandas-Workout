# Exercise 5 — Monday Temperatures

## Chapter 1: Series

### Problem

You have four weeks of daily temperature readings (°C) starting on a Monday.  
Create a `pandas.Series` with a `DatetimeIndex` and answer questions about Mondays.

```
# 28 daily readings starting 2024-01-01 (Monday)
temps = [5, 7, 6, 8, 9, 10, 8,
         6, 4, 5, 7, 9, 11, 10,
         8, 9, 10, 12, 14, 13, 11,
         9, 10, 11, 13, 15, 14, 12]
```

1. Build the Series with a daily `DatetimeIndex`.
2. Extract only the Monday readings.
3. What is the average Monday temperature?

## Working it out

````python
import pandas as pd

temps_data = [5, 7, 6, 8, 9, 10, 8,
              6, 4, 5, 7, 9, 11, 10,
              8, 9, 10, 12, 14, 13, 11,
              9, 10, 11, 13, 15, 14, 12]
````

## Solution

````python
import pandas as pd

temps_data = [5, 7, 6, 8, 9, 10, 8,
              6, 4, 5, 7, 9, 11, 10,
              8, 9, 10, 12, 14, 13, 11,
              9, 10, 11, 13, 15, 14, 12]

# Build Series with daily DatetimeIndex starting on a Monday
idx = pd.date_range(start='2024-01-01', periods=len(temps_data), freq='D')
temps = pd.Series(temps_data, index=idx)
print(temps)
````

````python
# Monday is day-of-week 0
mondays = temps[temps.index.day_of_week == 0]
print('Monday temperatures:')
print(mondays)
print(f'\nAverage Monday temperature: {mondays.mean():.1f} °C')
````

## Beyond the Exercise

- `DatetimeIndex.day_of_week` returns 0=Monday … 6=Sunday.
- `DatetimeIndex.day_name()` returns the full weekday name as strings — useful for labelling.
- Grouping by weekday: `temps.groupby(temps.index.day_of_week).mean()` gives the average for every day of the week at once.
