### Problem

You have a `pandas.Series` recording the number of passengers on a bus route for each trip over a week.  
The bus runs 5 trips per day, 7 days a week (35 trips total).

```
passengers = [
    12, 25, 30, 18, 8,    # Monday
    15, 28, 35, 20, 10,   # Tuesday
    10, 22, 28, 17, 7,    # Wednesday
    14, 27, 33, 19, 9,    # Thursday
    18, 32, 40, 25, 12,   # Friday
    20, 15, 10, 8, 5,     # Saturday
    10, 8, 6, 4, 3,       # Sunday
]
```

1. What is the total number of passengers over the week?
2. What is the average number of passengers per trip?
3. How many trips had fewer than 15 passengers?
4. What was the busiest trip (highest passenger count)?

## Working it out

````python
import pandas as pd

passengers = pd.Series([
    12, 25, 30, 18, 8,
    15, 28, 35, 20, 10,
    10, 22, 28, 17, 7,
    14, 27, 33, 19, 9,
    18, 32, 40, 25, 12,
    20, 15, 10, 8, 5,
    10, 8, 6, 4, 3,
])
````

## Solution

````python
import pandas as pd

passengers = pd.Series([
    12, 25, 30, 18, 8,
    15, 28, 35, 20, 10,
    10, 22, 28, 17, 7,
    14, 27, 33, 19, 9,
    18, 32, 40, 25, 12,
    20, 15, 10, 8, 5,
    10, 8, 6, 4, 3,
])

print(f'Total passengers      : {passengers.sum()}')
print(f'Average per trip      : {passengers.mean():.1f}')
print(f'Trips with < 15 pax   : {(passengers < 15).sum()}')
print(f'Busiest trip (index)  : {passengers.idxmax()} — {passengers.max()} passengers')
````
