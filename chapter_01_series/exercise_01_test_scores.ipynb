## Chapter 1: Series

### Problem

Create a `pandas.Series` from the following list of student test scores and answer some basic questions about it:

```
scores = [85, 92, 78, 90, 88, 76, 95, 83, 70, 91]
```

1. What is the maximum score?
2. What is the minimum score?
3. What is the mean (average) score?
4. How many students scored above 85?

## Working it out

````python
import pandas as pd

scores = [85, 92, 78, 90, 88, 76, 95, 83, 70, 91]
````

## Solution

````python
import pandas as pd

scores = pd.Series([85, 92, 78, 90, 88, 76, 95, 83, 70, 91])
print(scores)
````

````python
print(f'Maximum score : {scores.max()}')
print(f'Minimum score : {scores.min()}')
print(f'Mean score    : {scores.mean()}')
print(f'Above 85      : {(scores > 85).sum()}')
````

## Beyond the Exercise

- Use `scores.describe()` to get a full statistical summary in one call.
- Use `scores.value_counts()` to see how many times each score appears.
- Use boolean indexing (`scores[scores > 85]`) to retrieve the actual scores above 85.

````python
print(scores.describe())
````
