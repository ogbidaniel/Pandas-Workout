# Exercise 7 — Long, Medium, and Short Taxi Rides

## Chapter 1: Series

### Problem

You have a `pandas.Series` of taxi ride distances (miles) from New York City.  
Categorise each ride as **short** (< 2 miles), **medium** (2–10 miles), or **long** (> 10 miles).

```
distances = [0.5, 3.2, 12.1, 1.8, 7.4, 0.9, 15.6, 4.5, 2.0, 8.3,
             0.3, 11.0, 6.7, 1.2, 9.8, 0.7, 3.9, 14.2, 5.5, 2.8]
```

1. Create a new Series with labels `'short'`, `'medium'`, or `'long'` for each ride.
2. How many rides fall into each category?
3. What percentage of rides are long?

## Working it out

````python
import pandas as pd

distances = pd.Series([0.5, 3.2, 12.1, 1.8, 7.4, 0.9, 15.6, 4.5, 2.0, 8.3,
                       0.3, 11.0, 6.7, 1.2, 9.8, 0.7, 3.9, 14.2, 5.5, 2.8])
````

## Solution

````python
import pandas as pd

distances = pd.Series([0.5, 3.2, 12.1, 1.8, 7.4, 0.9, 15.6, 4.5, 2.0, 8.3,
                       0.3, 11.0, 6.7, 1.2, 9.8, 0.7, 3.9, 14.2, 5.5, 2.8])

# Categorise rides using pd.cut
categories = pd.cut(
    distances,
    bins=[0, 2, 10, float('inf')],
    labels=['short', 'medium', 'long'],
    right=True,
)

print('Ride categories:')
print(categories.values)
````

````python
counts = categories.value_counts().sort_index()
print('\nCount per category:')
print(counts)

long_pct = (categories == 'long').mean() * 100
print(f'\nPercentage of long rides: {long_pct:.1f}%')
````

## Beyond the Exercise

- `pd.cut()` creates a `Categorical` Series with ordered categories — useful for sorting and grouping.
- `pd.qcut()` is the quantile-based alternative that creates equally-sized groups by frequency.
- Combine with the original distances: `pd.DataFrame({'distance': distances, 'category': categories})` for richer analysis.
