## Chapter 1: Series

### Problem

Given a `pandas.Series` of raw test scores (out of 50), scale them to a range of 0–100:

```
raw_scores = [38, 45, 30, 42, 40, 28, 47, 35, 25, 44]
```

1. Multiply every score by 2 to convert to the 0–100 scale.
2. What is the new mean?
3. Add 5 bonus points to every score (cap at 100).

## Working it out

````python
import pandas as pd

raw_scores = pd.Series([38, 45, 30, 42, 40, 28, 47, 35, 25, 44])
````

## Solution

````python
import pandas as pd

raw_scores = pd.Series([38, 45, 30, 42, 40, 28, 47, 35, 25, 44])

# Scale to 0-100
scaled = raw_scores * 2
print('Scaled scores:')
print(scaled)
print(f'\nNew mean: {scaled.mean()}')
````

````python
# Add 5 bonus points, cap at 100
with_bonus = (scaled + 5).clip(upper=100)
print('Scores with 5-point bonus (capped at 100):')
print(with_bonus)
````

## Beyond the Exercise

- Pandas broadcasts arithmetic operations across the entire Series—no loops needed.
- `Series.clip(lower, upper)` is the idiomatic way to cap values.
- To normalise to a strict 0–1 range: `(s - s.min()) / (s.max() - s.min())`.
