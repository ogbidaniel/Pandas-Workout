# Exercise 3 — Counting Tens Digits

## Chapter 1: Series

### Problem

Generate a random `pandas.Series` of integer test scores (0–100), count how many scores fall into each "tens" bucket (0–9, 10–19, 20–29, … 90–99, 100).


1. Use integer division (`// 10`) to extract the tens digit of each score.
2. Count how many scores have each tens digit.
3. Display the counts sorted by tens digit.

## Working it out

```python
import pandas as pd
import numpy as np
```

## Solution

```python
s = np.random.randint(00,101,10)
s = pd.Series(s)
s
```

```python
# using div and converting type
(s / 10).astype(np.int8)
```

```python
s // 10
```

## Beyond the Exercise

- `pd.cut()` is another powerful way to bucket continuous data into labelled intervals.
- `value_counts(normalize=True)` gives proportions instead of raw counts.
- Chaining `.sort_index()` after `value_counts()` ensures a deterministic, ordered display.
