### Problem

Given a `pandas.Series` of daily temperatures (°C) recorded over two weeks, compute a full set of descriptive statistics.

```
temps = [22, 25, 19, 28, 30, 27, 24, 21, 26, 29, 23, 20, 18, 31]
```

1. Use `describe()` to get count, mean, std, min, quartiles, and max.
2. What is the variance?
3. What is the median temperature?
4. What percentage of days were hotter than 25 °C?

## Working it out

````python
import pandas as pd

temps = pd.Series([22, 25, 19, 28, 30, 27, 24, 21, 26, 29, 23, 20, 18, 31])
````

## Solution

````python
import pandas as pd

temps = pd.Series([22, 25, 19, 28, 30, 27, 24, 21, 26, 29, 23, 20, 18, 31])

print('Descriptive statistics:')
print(temps.describe())
````

````python
print(f'Variance : {temps.var():.2f}')
print(f'Median   : {temps.median()}')
pct_hot = (temps > 25).mean() * 100
print(f'Days > 25°C: {pct_hot:.1f}%')
````

## Beyond the Exercise

- `describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])` lets you customise the quantile rows.
- `Series.quantile(0.9)` gives the 90th percentile directly.
- Boolean mean (`(s > threshold).mean()`) is an idiomatic way to compute percentages.
