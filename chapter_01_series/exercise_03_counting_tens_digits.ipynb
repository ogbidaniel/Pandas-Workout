## Chapter 1: Series

### Problem

Given a `pandas.Series` of integer test scores (0–100), count how many scores fall into each "tens" bucket (0–9, 10–19, 20–29, … 90–99, 100).

```
scores = [85, 92, 78, 90, 88, 76, 95, 83, 70, 91, 64, 55, 48, 99, 100]
```

1. Use integer division (`// 10`) to extract the tens digit of each score.
2. Count how many scores have each tens digit.
3. Display the counts sorted by tens digit.

## Working it out

````python
import pandas as pd

scores = pd.Series([85, 92, 78, 90, 88, 76, 95, 83, 70, 91, 64, 55, 48, 99, 100])
````

## Solution

````python
import pandas as pd

scores = pd.Series([85, 92, 78, 90, 88, 76, 95, 83, 70, 91, 64, 55, 48, 99, 100])

# Extract tens digit using integer division
tens_digits = scores // 10
print('Tens digits:')
print(tens_digits.values)
````

````python
# Count occurrences of each tens digit, sorted by index
counts = tens_digits.value_counts().sort_index()
print('\nCounts per tens bucket:')
print(counts)
````

## Beyond the Exercise

- `pd.cut()` is another powerful way to bucket continuous data into labelled intervals.
- `value_counts(normalize=True)` gives proportions instead of raw counts.
- Chaining `.sort_index()` after `value_counts()` ensures a deterministic, ordered display.
