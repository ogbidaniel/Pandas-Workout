# Exercise 7 - Long, medium, and short taxi rides

Load the data from the taxi-distance.csv into a series. Modify (or create another series) the data from this series containing names rather than numbers using:

- Short, <= 2 miles
- Medium, > 2 miles
- Long, > 10 miles

```python
import pandas as pd
import numpy as np
```

```python
pathtodata = '/Users/ogbi/Development/Pandas-Workout/data/taxi-distance.csv'
s = pd.read_csv(pathtodata,
                header=None).squeeze()
s.head()
```

```python
# Using pd.cut to creat bins for our data and 
# count number of rides for each category
pd.cut(s,
       bins=[0, 2, 10, s.max()],
       include_lowest=True,
       labels=['short','medium','long']).value_counts()
```

## Beyond the exercise

```python
# Compare the mean and median trip distances. What does it say about the distribution
print(f'Mean: {s.mean()}')
print(f'Median: {s.median()}')
```
