```python
from pandas import Series, DataFrame
import numpy as np
import math
```

```python
# calculating the mean and standard deviation.
g = np.random.default_rng(seed=49)
scores = Series(g.integers(70,100,10))
scores
```

```python
math.sqrt(((scores - scores.mean()) ** 2).sum() / scores.count())
```

```python
scores.std(ddof=0)  # pass ddof=0 to use n instead of n - 1
```

Vectorized Operations in Pandas (and NumPy)

```python
s1 = Series([10,20,30,40])
s2 = Series([100,200,300,400])

s1 + s2
```

```python
s1 = Series([10,20,30,40], index=list('abcd'))
s2 = Series([100,200,300,400], index=list('dcba'))
s1 + s2
```
